Section A: Theory Questions

1. What is a random variable in probability theory?

Ans - A random variable is a numerical value assigned to each possible outcome of a random experiment. It is usually denoted by
a capital letter such as X. For example, if a coin is tossed three times, X could represent the number of heads. A random
variable lets us describe outcomes mathematically and assign probabilities to them.

2. What are the types of random variables?

Ans - The two main types are discrete and continuous. A discrete random variable takes distinct, countable values, such as the
number of calls received in an hour. A continuous random variable can take any value within an interval, such as height,
time, or temperature. Some courses also discuss mixed random variables, which combine discrete probability masses and
continuous density.

3. What is the difference between discrete and continuous distributions?

Ans - A discrete distribution assigns probabilities to individual values; its probability mass function (PMF) satisfies P(X = x) ³ 0 and
the probabilities over all possible values sum to 1. A continuous distribution is described by a probability density function
(PDF); probabilities are areas under the density curve, and P(X = x) = 0 for any single exact value. For a continuous
variable, P(a £ X £ b) is the integral of the density from a to b.

4. What are probability distribution functions (PDF)?

Ans - The term PDF usually means probability density function for a continuous random variable. A density f(x) is non-negative
and integrates to 1 over the full range. Probabilities are found by integration: P(a £ X £ b) = ònn f(x) dx. A density value itself
is not generally the probability at a point and can even exceed 1. For discrete variables, the corresponding function is called
a probability mass function (PMF).

5. How do cumulative distribution functions (CDF) differ from probability distribution functions
(PDF)?

Ans - A cumulative distribution function is F(x) = P(X £ x). It gives the accumulated probability up to x and is defined for both
discrete and continuous variables. For a continuous variable with density f, F(x) = òn¥n f(t) dt, and where differentiable, f(x)
= F¢(x). For a discrete variable, the CDF is a step function whose jumps equal the point probabilities.

6. What is a discrete uniform distribution?

Ans - A discrete uniform distribution gives equal probability to each value in a finite set. If X can take one of n values, then P(X = x)
= 1/n for each allowed value. For example, a fair six-sided die has P(X = k) = 1/6 for k = 1, 2, …, 6. Its expected value is the
average of the possible values when each is equally likely.

7. What are the key properties of a Bernoulli distribution?

Ans - A Bernoulli random variable represents one trial with two outcomes, conventionally success (1) and failure (0). If the
probability of success is p, then P(X = 1) = p and P(X = 0) = 1 - p. Its mean is E[X] = p and variance is Var(X) = p(1 - p). The
parameter p lies between 0 and 1 inclusive.

8. What is the binomial distribution, and how is it used in probability?

Ans - The binomial distribution counts the number of successes in n independent Bernoulli trials, each with the same success
probability p. Its probability mass function is P(X = k) = C(n,k)pn(1 - p)nnn, for k = 0, …, n. The mean is np and variance is
np(1 - p). It is useful for questions such as how many heads occur in repeated coin tosses or how many items in a batch are
defective.

9. What is the Poisson distribution and where is it applied?

Ans - The Poisson distribution models a count of events in a fixed interval of time, area, distance, or volume when events occur
independently at a constant average rate l under suitable conditions. Its PMF is P(X = k) = enl ln/k!, for k = 0, 1, 2, … . Its
mean and variance are both l. Examples include the number of calls arriving in a minute or defects per unit of material,
when the model assumptions are reasonable.

10. What is a continuous uniform distribution?

Ans - A continuous uniform distribution on [a, b] assigns constant density across the interval. Its PDF is f(x) = 1/(b - a) for a £ x £ b
and 0 otherwise. Its CDF rises linearly from 0 to 1 over that interval. The mean is (a + b)/2 and variance is (b - a)²/12.
Probabilities are proportional to interval length.

11. What are the characteristics of a normal distribution?

Ans - The normal distribution is continuous, bell-shaped, and symmetric about its mean m. Its mean, median, and mode are equal.
It is completely determined by its mean m and standard deviation s, with s > 0. The total area under its curve is 1, and its
tails extend indefinitely in both directions. Approximately 68%, 95%, and 99.7% of values lie within 1, 2, and 3 standard
deviations of the mean, respectively.

12. What is the standard normal distribution, and why is it important?

Ans - The standard normal distribution is a normal distribution with mean 0 and standard deviation 1, written Z ~ N(0, 1). It allows
values from different normal distributions to be converted to a common scale using Z = (X - m)/s. Standard normal tables or
software can then be used to calculate probabilities, percentiles, and critical values.

13. What is the Central Limit Theorem (CLT), and why is it critical in statistics?

Ans - The Central Limit Theorem states, under suitable conditions, that the distribution of the standardized sample mean
approaches a standard normal distribution as sample size increases, even when the population itself is not normal. For
independent identically distributed observations with finite mean m and finite variance s², the sample mean is approximately
normal for sufficiently large n, with mean m and standard error s/Ön. The CLT supports confidence intervals, hypothesis
tests, and approximations for sample averages.

14. How does the Central Limit Theorem relate to the normal distribution?

Ans - The CLT explains why the normal distribution frequently appears in statistical inference. When many independent
observations with finite variance are averaged or summed, their standardized aggregate tends toward a normal distribution
as sample size grows. This concerns the distribution of sample means or sums, not a claim that the original population
becomes normal. The sample size needed for a good approximation depends on the population’s shape and dependence
structure.

15. What is the application of Z statistics in hypothesis testing?

Ans - A Z statistic expresses how far a sample statistic is from a hypothesized value in standard-error units. Z tests are commonly
used for means when the population standard deviation is known, for sufficiently large-sample approximations, and for some
proportion tests when their conditions hold. The calculated Z value is compared with critical values or used to obtain a
p-value. A conclusion should be made using a preselected significance level and the test’s assumptions.

16. How do you calculate a Z-score, and what does it represent?

Ans - For an observation x from a population with mean m and standard deviation s, calculate z = (x - m)/s. A Z-score tells how
many standard deviations the observation is above (positive) or below (negative) the mean. For a sample-based
standardized score, the sample mean and sample standard deviation are often used. The interpretation depends on which
mean and standard deviation are appropriate for the problem.

17. What are point estimates and interval estimates in statistics?

Ans - A point estimate is one number used to estimate a population parameter; for example, the sample mean estimates the
population mean. An interval estimate gives a range of plausible values, usually with a stated confidence level. For example,
a 95% confidence interval is a procedure that, over repeated samples under its assumptions, would cover the true
parameter in about 95% of samples.

18. What is the significance of confidence intervals in statistical analysis?

Ans - A confidence interval communicates both an estimate and its sampling uncertainty. It provides a range of parameter values
compatible with the data and the chosen method, rather than presenting only a single estimate. Its width is affected by
sample size, variability, and confidence level. A 95% confidence interval does not mean that the fixed parameter has a 95%
probability of lying in this particular calculated interval under the standard frequentist interpretation.

19. What is the relationship between a Z-score and a confidence interval?

Ans - A confidence interval often takes the form estimate ± critical Z value × standard error when a normal-based method is
appropriate. For example, a two-sided 95% interval uses approximately 1.96 as the standard normal critical value. The Z
critical value determines how many standard errors are added on each side of the estimate. Other methods, such as t
intervals, are needed in situations where their assumptions are more suitable.

20. How are Z-scores used to compare different distributions?

Ans - Z-scores standardize observations by subtracting a distribution’s mean and dividing by its standard deviation. This
expresses each observation in standard-deviation units, making relative positions easier to compare across scales. A
Z-score of +2 is two standard deviations above its reference mean. Comparisons are most meaningful when the means and
standard deviations are estimated appropriately and the distributions’ shapes are considered.

21. What are the assumptions for applying the Central Limit Theorem?

Ans - Common conditions for the usual CLT include independent observations (or an appropriate weak-dependence condition), a
common distribution or suitable regularity conditions, and finite population variance. For an independent identically
distributed sample, finite mean and variance are standard assumptions. The approximation generally improves with larger
samples, but strongly skewed or heavy-tailed populations may require larger samples. Sampling without replacement from a
finite population may require a finite-population correction and attention to the sampling fraction.

22. What is the concept of expected value in a probability distribution?

Ans - The expected value is the probability-weighted long-run average of a random variable. For a discrete variable, E[X] = S x
P(X = x). For a continuous variable with density f(x), E[X] = ò x f(x) dx, when the expectation exists. It need not be a value the
variable can actually take. For example, the expected result of a fair six-sided die is 3.5.

23. How does a probability distribution relate to the expected outcome of a random variable?

Ans - A probability distribution describes which values a random variable can take and how likely they are. The expected value
combines those values with their probabilities to give a theoretical average over repeated trials. It is not a guarantee of the
outcome of any single trial. Other distribution features, such as variance, skewness, and tail probabilities, describe
uncertainty and shape beyond the mean.

```python
#26. Write a Python function to calculate the probability distribution function (PDF) of a Bernoulli distribution
def bernoulli_pmf(x, p):
    """Return P(X=x) for X ~ Bernoulli(p), where x is 0 or 1."""
    if not 0 <= p <= 1:
        raise ValueError("p must be between 0 and 1")
    if x == 1:
        return p
    if x == 0:
        return 1 - p
    return 0.0

p = 0.7
print("P(X=1) =", bernoulli_pmf(1, p))
print("P(X=0) =", bernoulli_pmf(0, p))
```

```python
#29. Write a Python program to calculate and plot the cumulative distribution function (CDF) of a discrete uniform distribution
import numpy as np
import matplotlib.pyplot as plt

values = np.arange(1, 7)
pmf = np.full(6, 1 / 6)
cdf = np.cumsum(pmf)

plt.step(values, cdf, where="post")
plt.scatter(values, cdf)
plt.xticks(values)
plt.yticks(np.arange(0, 1.1, 1/6))
plt.xlabel("x")
plt.ylabel("F(x) = P(X <= x)")
plt.title("CDF of a Discrete Uniform Distribution")
plt.ylim(0, 1.05)
plt.grid(True, alpha=0.3)
plt.show()
print("CDF values:", cdf)
```

```python
#38. Implement hypothesis testing using Z-statistics
import numpy as np
from scipy.stats import norm

# Example: H0: population mean = 50; known population SD = 10
data = np.array([52, 48, 51, 55, 49, 53, 50, 54, 47, 51], dtype=float)
mu0, sigma, alpha = 50, 10, 0.05
n = len(data)
z_stat = (data.mean() - mu0) / (sigma / np.sqrt(n))
p_value = 2 * (1 - norm.cdf(abs(z_stat)))  # two-sided test

print("Sample mean:", data.mean())
print("Z statistic:", z_stat)
print("Two-sided p-value:", p_value)
if p_value < alpha:
    print("Reject H0 at the 5% significance level.")
else:
    print("Fail to reject H0 at the 5% significance level.")

# Assumptions: observations are independent and population SD is known;
# normal population or an adequate large-sample approximation.
```

```python
#39. Create a confidence interval for a dataset using Python and interpret the result
import numpy as np
from scipy.stats import t

data = np.array([12, 15, 14, 10, 18, 20, 13, 16], dtype=float)
n = len(data)
mean = data.mean()
s = data.std(ddof=1)
confidence = 0.95
critical_t = t.ppf((1 + confidence) / 2, df=n - 1)
margin = critical_t * s / np.sqrt(n)
lower, upper = mean - margin, mean + margin

print(f"Sample mean: {mean:.3f}")
print(f"95% t confidence interval: ({lower:.3f}, {upper:.3f})")
print("Interpretation: this method produces intervals that cover the true population mean")
print("in about 95% of repeated samples, under its assumptions.")
```

```python
#40. Generate data from a normal distribution, then calculate and interpret the confidence interval for its mean
import numpy as np
from scipy.stats import t

rng = np.random.default_rng(42)
data = rng.normal(loc=100, scale=15, size=100)
mean = data.mean()
s = data.std(ddof=1)
n = len(data)
tcrit = t.ppf(0.975, df=n - 1)
ci = (mean - tcrit * s / np.sqrt(n), mean + tcrit * s / np.sqrt(n))

print("Sample mean:", round(mean, 3))
print("95% confidence interval for the mean:", tuple(round(v, 3) for v in ci))
```

```python
#46. Implement Z-score normalization and visualize the transformation
import numpy as np
import matplotlib.pyplot as plt

data = np.array([12, 15, 14, 10, 18, 20, 13, 16], dtype=float)
mean = data.mean()
std = data.std(ddof=0)
z = (data - mean) / std

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].hist(data, bins=6, edgecolor="black")
axes[0].set_title("Original Data")
axes[0].set_xlabel("Value")
axes[1].hist(z, bins=6, edgecolor="black")
axes[1].set_title("Z-score Normalized Data")
axes[1].set_xlabel("Z-score")
plt.tight_layout()
plt.show()

print("Normalized mean:", z.mean())
print("Normalized population SD:", z.std(ddof=0))
```
